```python
import pandapower as pp
net = pp.create_empty_network()

# create bus
b1 = pp.create_bus(net,vn_kv=.4, name='bus 1')
b2 = pp.create_bus(net,vn_kv=.4, name='bus 2')
b3 = pp.create_bus(net,vn_kv=.4, name='bus 3')
b4 = pp.create_bus(net,vn_kv=.4, name='bus 4')

#slack bus
pp.create_ext_grid(net,b1,vm_pu=1.08,name="slack")

# create tran
#pp.create_transformer(net,b1,b2, std_type="0.4 MVA 20/0.4 kV") 

pp.create_transformer_from_parameters(net, hv_bus=b1, lv_bus=b2, name="Tranfo1", 
                                      sn_mva=10, vn_hv_kv=.4, vn_lv_kv=.4, vk_percent=1,
                                      vkr_percent=0.00995, pfe_kw=0, i0_percent=0,tap_side='lv',tap_phase_shifter=False,
                                      tap_step_percent=0.625,tap_max=16,tap_min=-16,tap_neutral=0,tap_pos=0)


#create line
pp.create_line(net,b2,b3,std_type="NAYY 4x50 SE", length_km=0.1, name="line1")
pp.create_line(net,b3,b4,std_type="NAYY 4x50 SE", length_km=0.1, name="line2")
#create load
pp.create_load(net,b4,p_mw=0.1,q_mvar=0.05)
```

```python
import pandapower.control as control

trafo_controller=control.LineDropControl(net=net,tid=0,vm_lower_pu=0.99, vm_upper_pu=1.01, vm_set_pu_val=1.0,CT=500,PT=1.92,R_comp=12.0,X_comp=1.6,loc_sen=False)
```

```python
pp.runpp(net, run_control=True)
```
